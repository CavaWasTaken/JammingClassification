**Test Features**

```python
from google.colab import drive
from pathlib import Path

drive.mount('/content/drive')

path_to_csv = Path('./drive/MyDrive/N-MON/Processed_Dataset/FEATURES')
```

```python
import pandas as pd

for csv_file in sorted(path_to_csv.glob('*.csv')):
    # load the csv file of the signal
    signal_data = pd.read_csv(csv_file) 

    # print csv column names
    print(f"Columns in {csv_file.name}: {signal_data.columns.tolist()}")

    # print number of rows and columns in the dataframe
    print(f"Shape of dataframe from {csv_file.name}: {signal_data.shape}")

    # print the first few rows of the dataframe
    print(f"Data from {csv_file.name}:")
    print(signal_data.head())
    print("\n")

    # print the last few rows of the dataframe
    print(f"Last rows of {csv_file.name}:")
    print(signal_data.tail())
    print("\n")

    # check for missing values in the dataframe
    missing_values = signal_data.isnull().sum()
    # find the index of the rows with missing values
    missing_rows = signal_data[signal_data.isnull().any(axis=1)].index.tolist()
    print(f"Missing values in {csv_file.name}:\n{missing_values}")
    print(f"Rows with missing values in {csv_file.name}: {missing_rows}\n")
    # show the value contained in the rows with missing values
    if missing_rows:
        print(f"Values in rows with missing values in {csv_file.name}:")
        print(signal_data.loc[missing_rows])
        print("\n")
    print("-" * 50 + "\n")
```

```python
# remove P1 samples from the csv
for csv_file in sorted(path_to_csv.glob('*.csv')):
    # load the csv file of the signal
    signal_data = pd.read_csv(csv_file) 

    # if the power column exists
    if 'power' not in signal_data.columns:
        print(f"Column 'power' not found in {csv_file}")
        continue

    # remove rows where the power column contains 'P1'
    signal_data = signal_data[signal_data['power'] != 'P1']

    # save the modified dataframe back to a new csv file
    modified_csv_file = csv_file.with_name(f"{csv_file.stem}_modified.csv")
    signal_data.to_csv(modified_csv_file, index=False)
    print(f"Modified CSV saved as: {modified_csv_file}\n")
```
