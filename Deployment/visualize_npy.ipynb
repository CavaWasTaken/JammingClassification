```python
# read npy matrix of the spectogram
import matplotlib
matplotlib.use('Agg')  # Set backend before importing pyplot
import matplotlib.pyplot as plt
import numpy as np

npy_file = 'spec_1777050100.1657455.npy'

spectrogram = np.load(npy_file)
# write the spectogram into a readable text file
np.savetxt('spectrogram_output.txt', spectrogram, fmt='%.6f')
```
