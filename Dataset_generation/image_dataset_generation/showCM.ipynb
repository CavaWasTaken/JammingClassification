```python
# load the confusion matrix saved in csv file

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

cm = pd.read_csv(matrices_dir + '/confusion_matrix_ResNet18_41Classes.csv', index_col=0)
cm = cm.to_numpy()

class_names = ['CLEAN', 'LN_P1', 'LN_P2', 'LN_P3', 'LN_P4', 'LN_P5', 'LN_P6', 'LN_P7', 'LN_P8', 
               'LWF_P1', 'LWF_P2', 'LWF_P3', 'LWF_P4', 'LWF_P5', 'LWF_P6', 'LWF_P7', 'LWF_P8', 
               'TICK_P1', 'TICK_P2', 'TICK_P3', 'TICK_P4', 'TICK_P5', 'TICK_P6', 'TICK_P7', 'TICK_P8',
               'TRI_P1', 'TRI_P2', 'TRI_P3', 'TRI_P4', 'TRI_P5', 'TRI_P6', 'TRI_P7', 'TRI_P8',
               'TRIW_P1', 'TRIW_P2', 'TRIW_P3', 'TRIW_P4', 'TRIW_P5', 'TRIW_P6', 'TRIW_P7', 'TRIW_P8']


# plot the confusion matrix
plt.figure(figsize=(20, 17))
sns.heatmap(cm, annot=True, fmt='g', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix')
plt.show()
```
