**Convert the dataset from float64 to int8 in Python using NumPy**

```python
import os
import numpy as np
```

```python
def convert_float64_to_int8(matrix_npy):
    matrix_int8 = ((matrix_npy - matrix_npy.min()) / (matrix_npy.max() - matrix_npy.min()) * 255 - 128).astype(np.int8)
    return matrix_int8
```

```python
from google.colab import drive, files
drive.mount('/content/drive')
dataset_path = '/content/drive/MyDrive/N-MON/ALTRO/spectograms/MULTITONE-NARROW-40-20'
save_path = '/content/drive/MyDrive/N-MON/Processed_Dataset/MULTITONE-NARROW-40-20/'

# if the save path folder doesn't exist, create it
os.makedirs(save_path, exist_ok=True)

folder_NoPower = ['CLEAN', 'anntennina2', 'MULTITONE-NARROW-40-20', 'LINEAR-WIDE-20', 'FH-20', 'HOOKED-SAWTOOTH-3-20']
```

```python
# iterate over each folder in the dataset directory, representing different signal types
# for signal_folder in os.listdir(dataset_path):
signal_folder = dataset_path
print(f'Processing signal type folder: {signal_folder}')
if os.path.isdir(signal_folder):
    # if the folder is not 'CLEAN', go one level deeper since it contains power subfolders
    if signal_folder.split('/')[-1] not in folder_NoPower:
        for power_folder in os.listdir(signal_folder):
            print(f'  Processing power subfolder: {power_folder}')
            power_folder_path = os.path.join(signal_folder, power_folder)  # path to the power subfolder
            if os.path.isdir(power_folder_path):
                # for each file inside the power subfolder
                for file in os.listdir(power_folder_path):
                    if file.endswith('.npy'):
                        file_path = os.path.join(power_folder_path, file)
                        # load the float64 matrix
                        matrix_float64 = np.load(file_path)
                        # convert to int8
                        matrix_int8 = convert_float64_to_int8(matrix_float64)
                        # save the int8 matrix
                        save_folder  = f'{save_path}/{power_folder}'
                        os.makedirs(save_folder, exist_ok=True)
                        save_file = os.path.join(save_folder, file)
                        print(f'  Saving converted file: {save_file}')
                        np.save(save_file, matrix_int8)

    # if the folder is 'CLEAN', process files directly inside it
    else:
        # for each file inside the subfolder
        for file in os.listdir(signal_folder):
            if file.endswith('.npy'):
                print(f'  Processing file: {file}')
                file_path = os.path.join(signal_folder, file)
                # load the float64 matrix
                matrix_float64 = np.load(file_path)
                # convert to int8
                matrix_int8 = convert_float64_to_int8(matrix_float64)
                # save the int8 matrix
                os.makedirs(save_path, exist_ok=True)
                save_file = os.path.join(save_path, file)
                np.save(save_file, matrix_int8)
                print(f'  Saved converted file: {save_file}')
```
